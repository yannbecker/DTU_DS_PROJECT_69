## Implement LSH 

### Obtain optimal value of b for given r and t

```python
from math import pow
print(pow(2,3))
```

```python
from math import floor, ceil

def b_optimal(k,t):

    def f(x,k,t):
        return x*(pow(t,k/x))-1
    binf = 1
    bsup = 1000
    bmax = bsup
    bmin = binf
    while bmax-bmin > 1 :
        bmoy = (bmax + bmin) / 2
        
        if f(bmoy,k,t)*f(bmax,k,t)<0:
            bmin = floor(bmoy)
        elif f(bmoy,k,t)*f(bmin,k,t)<0:
            bmax = ceil(bmoy)
        else :
            message =f"optimal value for b not between {binf} and {bsup} with k = {k}"
            class BadValue(Exception):
                """Exception levée pour des erreurs spécifiques dans notre application."""
                def __init__(self, message="Une erreur personnalisée est survenue."):
                # Appelle le constructeur de la classe parente (Exception)
                    super().__init__(message)
                    self.message = message
            raise BadValue(message)
    
    if abs(f(bmin,k,t)) > abs(f(bmax,k,t)):
        return bmax
    else :
        return bmin
        
```

```python
b = 4
r = 5
k = b*r # Signature length
q = 7 # Number of caracters per shingle
```

```python
docs = {} #dictionary mapping document id to document contents

# read data sets
srcfolder = os.path.dirname(os.path.abspath("Tests of the different functions.ipynb"))
datafolder = os.path.join(srcfolder, "ats_corpus_small")   # change to ats_corpus for large data set

for file in os.listdir(datafolder):
    filepath = os.path.join(datafolder, file)
    f = open(filepath, 'r')
    docs[file] = f.read()
    print("read document " + file)
    f.close()
```

```python
dic_signatures = signatures(docs, q, k)
```

```python
doc_ids = list(dic_signatures.keys())
id0 = doc_ids[0] # Input for the lsh research
```

```python
similar_candidates = {}
n = len(dic_signatures)
m = 2*n # number of buckets
```

```python
from lsh_utils.Bucket_hashing import lsh_band_hash    
```

```python
for band_nb in range(b):
    input_hash = lsh_band_hash(
        band = dic_signatures[id0][band_nb*r:(band_nb+1)*r],
        m = m,
        lsh_seed = band_nb
        )
    for doc_nb in range(0,n):
        doc_id = doc_ids[doc_nb]
        doc_hash = lsh_band_hash(
            band = dic_signatures[doc_id][band_nb*r:(band_nb+1)*r],
            m = m,
            lsh_seed = band_nb
            )
        if doc_hash == input_hash :
            if doc_id in similar_candidates :
                similar_candidates[doc_id] += 1
            else :
                similar_candidates[doc_id] = 1
            
```
